```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
import csv

df = pd.read_csv('/content/marketing_campaign_dataset.csv', quoting=csv.QUOTE_NONE)
```

```python
df.head(2)
```

```python
print('Duplicate Row:', df.duplicated().sum())

print('\nMissing values:')
print(df.isnull().sum())

print('\nData Types:')
print(df.dtypes)

# We check for the column names. Depending on quoting, they might have leading/trailing whitespaces or quotes.
print('\nAvailable Columns:', list(df.columns))

# Safely show samples if the columns exist
for col in ['Acquisition_Cost', 'Date', 'Campaign_ID']:
    if col in df.columns:
        print(f'\n{col} sample:')
        print(df[col].head(5))
    else:
        # Check if they exist with quotes or whitespaces
        matching_cols = [c for c in df.columns if col in c]
        if matching_cols:
            print(f'\nFound matching column name "{matching_cols[0]}" instead of "{col}":')
            print(df[matching_cols[0]].head(5))

print('\nTotal rows:', len(df))
```

```python
missing_rows = df[df.isnull().any(axis=1)]

print('Rows with missing values:', len(missing_rows))
display(missing_rows)
```

```python
print("Dataset shape:", df.shape)

print("\nMissing values per column:")
print(df.isnull().sum())

print("\nTotal missing values:", df.isnull().sum().sum())

print("\nDuplicate rows:", df.duplicated().sum())
```

```python
print(df["Acquisition_Cost"].dtype)
print(df["Acquisition_Cost"].head())

print(df["Date"].dtype)
print(df["Date"].head())

print(df["Campaign_ID"].nunique())
```

```python
df_clean = df.copy()

df_clean['Acquisition_Cost'] = pd.to_numeric(df['Acquisition_Cost'].str.replace("$", "", regex=False).str.replace(",", "", regex=False),errors="coerce")
df_clean['Date'] = pd.to_datetime(df['Date'],format = "%d/%m/%Y", errors="coerce")


print(df_clean["Acquisition_Cost"].dtype)
print(df_clean["Date"].dtype)


print(df_clean[["Acquisition_Cost", "Date"]].isnull().sum())

display(df_clean.head())
```

```python
print("Original date samples:")
print(df["Date"].head(10).tolist())

print(df["Date"].dtype)

print("\nDate examples from cleaned data:")
print(df_clean["Date"].head(10))

print("\nDate values that failed conversion:")
print(df.loc[df_clean["Date"].isna(), "Date"].head(10).tolist())
```

```python
# Convert dates using automatic format inference
df_clean["Date"] = pd.to_datetime(
    df["Date"].astype("string").str.strip(),
    errors="coerce"
)

# Verify
print("Date data type:", df_clean["Date"].dtype)
print("Missing dates:", df_clean["Date"].isna().sum())

print(df_clean["Date"].head())
```

```python
# Check numeric column data types
print(df_clean.dtypes)

# Check statistical ranges
display(
    df_clean[
        [
            "Conversion_Rate",
            "Acquisition_Cost",
            "ROI",
            "Clicks",
            "Impressions",
            "Engagement_Score"
        ]
    ].describe()
)
```

```python
df_clean["Conversion_Rate"] = pd.to_numeric(
    df_clean["Conversion_Rate"].astype(str).str.replace('"', '', regex=False).str.strip(),
    errors="coerce"
)

print("Conversion rates outside 0-1:",
      (~df_clean["Conversion_Rate"].between(0, 1)).sum())

print("Negative acquisition costs:",
      (df_clean["Acquisition_Cost"] < 0).sum())

print("Negative ROI values:",
      (df_clean["ROI"] < 0).sum())

print("Clicks greater than impressions:",
      (df_clean["Clicks"] > df_clean["Impressions"]).sum())

print("Invalid engagement scores:",
      (~df_clean["Engagement_Score"].between(1, 10)).sum())

print("\nDate range:")
print("Start:", df_clean["Date"].min())
print("End:", df_clean["Date"].max())

print("\nFinal shape:", df_clean.shape)
print("Missing values:", df_clean.isnull().sum().sum())
print("Duplicate rows:", df_clean.duplicated().sum())
```

```python
print("Export successful!")
print("Rows exported:", len(df_clean))
print("Columns exported:", len(df_clean.columns))
```
